```python
# Import Library

import matplotlib.pyplot as plt
import numpy as np
```

# Visualisasi Perbandingan ROUGE Test

```python

# Skenario Tokenizer
scenarios = ['baseline', 'indomixedtufs4', 'detiknews-pemilu', 'mix']

# Rata-rata skor ROUGE-1 Tes
rouge1_test_scores = [16.26, 24.56, 25.36, 25.40]
# Rata-rata skor ROUGE-2 Tes
rouge2_test_scores = [7.91, 12.41, 12.95, 12.85]

x = np.arange(len(scenarios))
width = 0.40  # lebar batang

fig, ax = plt.subplots(figsize=(9, 5)) # Atur ukuran gambar

# Membuat batang untuk ROUGE-1 Tes
rects1 = ax.bar(x - width/2, rouge1_test_scores, width, label='ROUGE-1 (Tes)')
# Membuat batang untuk ROUGE-2 Tes
rects2 = ax.bar(x + width/2, rouge2_test_scores, width, label='ROUGE-2 (Tes)')

# Menambahkan label, judul, dan tick labels
# ax.set_ylabel('Rata-Rata Skor')
# ax.set_xlabel('Skenario Tokenizer')
ax.set_title('Rata-Rata Skor ROUGE pada Set Tes', pad=15)
ax.set_xticks(x)
ax.set_xticklabels(scenarios)
ax.legend()

# Fungsi untuk menambahkan label nilai di atas setiap batang
def autolabel(rects, ax_plot):
    for rect in rects:
        height = rect.get_height()
        ax_plot.annotate(f'{height:.2f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=10)

autolabel(rects1, ax)
autolabel(rects2, ax)

# fig.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.7)
# Menyesuaikan batas atas sumbu Y agar label tidak terpotong
y_max_rouge = max(max(rouge1_test_scores), max(rouge2_test_scores))
ax.set_ylim(0, 100)


plt.show()
```

# Perbandingan Visualisasi BERTScore Test

```python
import matplotlib.pyplot as plt
import numpy as np

# Skenario Tokenizer
scenarios = ['baseline', 'indomixedtufs4', 'detiknews', 'mix']

# Rata-rata skor BERTScore Tes
bertscore_test_scores = [68.70, 71.23, 71.39, 71.32]

x = np.arange(len(scenarios))  # lokasi untuk skenario
width = 0.6  # lebar batang bisa lebih besar karena hanya satu set data per skenario

fig, ax = plt.subplots(figsize=(9, 6)) # Atur ukuran gambar

# Membuat batang untuk BERTScore Tes
rects_bert = ax.bar(x, bertscore_test_scores, width, label='BERTScore (Tes)')

# Menambahkan label, judul, dan tick labels
# ax.set_ylabel('Rata-Rata Skor BERTScore')
# ax.set_xlabel('Skenario Tokenizer')
ax.set_title('Rata-Rata Skor BERTScore pada Set Tes', pad=15)
ax.set_xticks(x)
ax.set_xticklabels(scenarios)
# ax.legend()

# Fungsi untuk menambahkan label nilai di atas setiap batang
def autolabel(rects, ax_plot):
    for rect in rects:
        height = rect.get_height()
        ax_plot.annotate(f'{height:.2f}',
                         xy=(rect.get_x() + rect.get_width() / 2, height),
                         xytext=(0, 3),  # 3 points vertical offset
                         textcoords="offset points",
                         ha='center', va='bottom', fontsize=11)

autolabel(rects_bert, ax)

# fig.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.7)
# Menyesuaikan batas bawah dan atas sumbu Y agar lebih fokus ke perbedaan skor
ax.set_ylim(0, 100) 

# Untuk menyimpan gambar (ganti nama file jika perlu)
# plt.savefig('gambar_4_Z_perbandingan_skor_bertscore.png', dpi=300)
```
